```python
import gurobipy as gp

gurobi_params = get_gurobi_params()
with grp.Env(params=gurobi_params) as env:
    # Load the model from your .lp file
    dhov = gp.read("dhov.lp", env)
    snv = gp.read("snv+milp.lp", env)

    # Optimize it
    dhov.optimize()
    snv.optimize()
```

```python
def var_names_to_txt(model, file_name):
    var_names = sorted([v.VarName for v in model.getVars()])

    with open(file_name, "w") as f:
        for name in var_names:
            f.write(f"{name}\n")

    print(f"Successfully wrote {len(var_names)} variable names to {file_name}")
```

```python
var_names_to_txt(dhov, "dhov_var_names.txt")
var_names_to_txt(snv, "snv_var_names.txt")
```
